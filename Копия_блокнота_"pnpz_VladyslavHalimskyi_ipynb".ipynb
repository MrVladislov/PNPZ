# **Widzenie maszynowe. Biblioteka OpenCV**

### Historia rozwoju widzenia maszynowego  
Widzenie maszynowe, będące połączeniem technologii przetwarzania obrazu, sztucznej inteligencji i automatyki, przeszło znaczną ewolucję od początkowych koncepcji teoretycznych po współczesne zaawansowane systemy. Poniżej przedstawiono kluczowe etapy jego rozwoju:

---

#### **1. Lata 40.–50. XX wieku: Początki teoretyczne**  
- **1940–1950**: Badania nad sztuczną inteligencją (SI) zainicjowane przez armię USA, które obejmowały analizę obrazu do celów wojskowych .  
- **1950**: James Gibson wprowadził teorię „przepływu optycznego”, która stała się podstawą analizy ruchu w obrazach .  
- **1956**: Termin „sztuczna inteligencja” został formalnie zdefiniowany przez Johna McCarthy’ego, co zapoczątkowało interdyscyplinarne badania .
48-John-McCarthy-AP (2).png
---

#### **2. Lata 60.–70. XX wieku: Narodziny koncepcji**  
- **1960**: Larry Roberts zaproponował możliwość wyodrębniania informacji 3D z obrazów 2D, co dało podwaliny pod trójwymiarowe widzenie maszynowe .  
- **1966**: Marvin Minsky z MIT zlecił stworzenie systemu łączącego kamerę z komputerem, co zaowocowało pierwszymi matematycznymi modelami przetwarzania obrazu .  
- **1978**: David Marr opracował teorię przetwarzania obrazu w trzech etapach (obliczeniowym, algorytmicznym, implementacyjnym), która stała się kluczowa dla rozwoju systemów wizyjnych .

375px-Larry_Roberts.jpg
---

#### **3. Lata 80. XX wieku: Industrializacja**  
- **Rozwój przemysłowy**: Powstały firmy specjalizujące się w technologiach wizyjnych, takie jak **Cognex**, **Basler** i **DALSA**, które wprowadziły pierwsze komercyjne systemy do kontroli jakości w fabrykach .  
- **Technologie**: Wprowadzono metody analizy obrazu oparte na piramidach obrazów oraz algorytmy do lokalizacji obiektów w przestrzeni 3D (np. *shape from shading*) .  
- **OCR**: Systemy optycznego rozpoznawania znaków zaczęto stosować w przemyśle do odczytu kodów i tekstów .

---

#### **4. Lata 90. XX wieku: Era inteligentnych kamer**  
- **Postęp sprzętowy**: Komputery i kamery stały się wystarczająco wydajne, by przetwarzać obrazy w czasie rzeczywistym. Powstały „inteligentne kamery” z wbudowanymi procesorami, które eliminowały konieczność użycia zewnętrznych komputerów .  
- **Algorytmy 3D**: Rozwój technik modelowania obiektów 3D przy użyciu stereoskopii, analizy cieni i natężenia światła .

---

#### **5. XXI wiek: Integracja z AI i ekspansja**  
- **2000–2010**: Wprowadzenie modułu **DNN** (Deep Neural Networks) w OpenCV, co umożliwiło integrację z modelami głębokiego uczenia .  
- **2010–2020**: Rozwój systemów opartych na **uczeniu maszynowym** i **sieciach neuronowych**, które zastąpiły tradycyjne algorytmy oparte na regułach .  
- **Przemysł 4.0**: Widzenie maszynowe stało się kluczowe w automatyzacji linii produkcyjnych, np. do kontroli jakości, pozycjonowania robotów czy identyfikacji kodów kreskowych .  
- **Chiny**: Od 2016 r. kraj ten stał się globalnym liderem w produkcji systemów wizyjnych, z firmami takimi jak **OPT** i **Daheng Images** .

---

#### **6. Współczesność (po 2020 r.): Rewolucja AI i przyszłość**  
- **Głębokie uczenie**: Modele takie jak **YOLO** (You Only Look Once) czy **EfficientNet** zrewolucjonizowały detekcję obiektów i segmentację obrazów .  
- **Multimodalność**: Integracja czujników wizyjnych z lidarami, czujnikami podczerwieni i dźwiękowymi dla pełniejszej percepcji środowiska .  
- **Robotyka**: Humanoidalne roboty, takie jak **Pepper** i **Nao**, wykorzystują widzenie maszynowe do interakcji z ludźmi i nawigacji w przestrzeni .  
- **Medycyna i logistyka**: Zastosowania w diagnostyce obrazowej, sortowaniu paczek czy autonomicznych magazynach .

### Podsumowanie  
Widzenie maszynowe ewoluowało od teoretycznych koncepcji do technologii przekształcającej przemysł, medycynę i robotykę. ---

### **OpenCV 1.0** (2006)  
- **Pierwsza stabilna wersja** biblioteki, zaprojektowana głównie do badań akademickich.  
- **Podstawowe funkcje**: detekcja krawędzi, filtracja obrazu, kalibracja kamery.  
- **Interfejs**: C API (bez wsparcia dla C++).  

---

### **OpenCV 2.x** (2008–2017)  
#### **Wersja 2.4.x (LTS)**  
- **Moduł `features2d`**: Wprowadzenie algorytmów **SIFT**, **SURF**, **ORB** do wykrywania i dopasowywania cech.  
- **GPU Acceleration**: Wsparcie dla CUDA (moduł `gpu`), przyspieszające operacje na kartach NVIDIA.  
- **OpenCL**: Wstępna integracja z obliczeniami heterogenicznymi.  
- **Detektory obiektów**: Algorytm **Haar Cascades** (używany m.in. do detekcji twarzy).  
- **Python API**: Pierwsze oficjalne wsparcie dla Pythona 2.x.  

#### **Wersja 2.3.1** (2011)  
- **Nowy interfejs C++**: Stopniowe zastępowanie C API.  
- **Moduł `ml`**: Algorytmy uczenia maszynowego (SVM, Random Forests).  

---

### **OpenCV 3.x** (2015–2018)  
#### **Wersja 3.0** (2015)  
- **Moduł `DNN` (beta)**: Wsparcie dla wczytywania modeli **Caffe** i **TensorFlow**.  
- **Transparent API (T-API)**: Automatyczne wykorzystanie OpenCL do przyspieszenia operacji.  
- **Usunięcie starego kodu**: Wycofanie przestarzałych funkcji C API (np. `cvLoadImage`).  
- **Moduł `viz`**: Wizualizacja 3D (np. chmury punktów).  

#### **Wersja 3.4.x** (2018)  
- **Detektory cech**: **AKAZE**, **BRISK** oraz wsparcie dla **Deep Learning** w `DNN`.  
- **Optical Flow**: Algorytm **DIS** (Dense Inverse Search).  
- **SIFT i SURF**: Przywrócone po usunięciu w 3.0 (wymagana flaga `OPENCV_ENABLE_NONFREE`).  

---

### **OpenCV 4.x** (2018–obecnie)  
#### **Wersja 4.0** (2018)  
- **C++11 jako domyślny standard**: Usunięto wsparcie dla C++98.  
- **G-API**: Nowy moduł do tworzenia grafów przetwarzania obrazu (optymalizacja wielowątkowości).  
- **Usunięcie C API**: Całkowite wycofanie starych funkcji (np. `cvQueryFrame`).  
- **DNN**: Wsparcie dla **ONNX**, **TensorFlow 1.x**, i **Darknet** (YOLOv3).  

#### **Wersja 4.5.x** (2020–2021)  
- **QR Code Detector**: Algorytm do wykrywania i dekodowania kodów QR.  
- **Moduł `rapid`**: Szybka detekcja markerów typu AprilTag.  
- **Optymalizacje NEON/AVX**: Przyspieszenie operacji na ARM i x86.  
- **Wsparcie dla ROS2**: Integracja z Robot Operating System.  

#### **Wersja 4.7.0** (2022)  
- **Moduł `wechat_qrcode`**: Detektor kodów QR od WeChat (większa dokładność).  
- **DeepLabV3+**: Wsparcie dla segmentacji semantycznej w `DNN`.  
- **Wsparcie dla ONNX opset 16**: Lepsza kompatybilność z modelami PyTorch.  

---

### **OpenCV 4.8.0** (2023)  
- **TFLite**: Pierwsze pełne wsparcie dla modeli **TensorFlow Lite** (w tym INT8).  
- **FFmpeg 5.0**: Aktualizacja dekodowania wideo (HDR, HEVC).  
- **CUDA 12.0**: Optymalizacje dla kart NVIDIA Ampere.  

---

### **OpenCV 4.9.0** (LTS, 2023)  
- **Winograd FP16**: Przyspieszenie operacji konwolucyjnych na ARM.  
- **Moduł `objdetect`**: Lepsza detekcja znaczników ArUco.  
- **Optymalizacje RISC-V**: Wstępne wsparcie dla architektury open-source.  

---

### **OpenCV 4.10.0** (2024)  
- **YOLOv10**: Integracja z najnowszymi modelami YOLO.  
- **NumPy 2.0**: Eksperymentalne wsparcie w Pythonie.  
- **Qualcomm HAL**: Optymalizacje dla Snapdragon.  

---

### **OpenCV 4.11.0** (2025)  
- **C++20**: Eksperymentalne wykorzystanie konceptów i `<format>`.  
- **JPEG XL i AVIF**: Nowe formaty kompresji obrazu.  
- **ONNX Runtime 1.17**: Wsparcie dla inferencji na Intel Arc i AMD ROCm.  

---

### **OpenCV 5.0** (planowane)  
- **C++17 jako minimum**: Wymóg nowoczesnych kompilatorów.  
- **Usunięcie C API**: Ostateczne pożegnanie z interfejsem w stylu C.  
- **G-API 2.0**: Grafowe przetwarzanie strumieniowe w czasie rzeczywistym.
